```python
import requests
from bs4 import BeautifulSoup
import csv
```

```python
# get data at url
content = requests.get('https://www.aha.org/fact-sheets/2025-06-05-medicaid-spending-reductions-would-lead-losses-jobs-economic-activity-and-tax-revenue-states').content
```

```python
# add a webscraping tool
soup = BeautifulSoup(content,'html.parser')
```

```python
# define the variables
table = soup.find('table')
header = table.find('thead')
```

```python
# define the variable and the information you are searching for
headers = header.find_all('th')
headers
```

```python
# define the outfile name and label what the csv will be called
outfile = open('econ_impacts.csv','w')
outcsv = csv.writer(outfile)

# write header first
outcsv.writerow(headers)

# find the table rows
rows = table.find_all('tr')

# 50 states
for row in rows[1:]:
  data = row.find_all('td')
  print(data)
  outcsv.writerow(data)
```

```python
# the end of the table you want
outfile.close()
```
